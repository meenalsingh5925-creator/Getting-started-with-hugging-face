```python
import pandas as pd

feedback_data = {
    "feedback": [
        "The course content was very well structured.",
        "The lectures were easy to understand.",
        "I really enjoyed the practical sessions and assignments.",
        "The instructor explained concepts clearly.",
        "The workload was heavy and deadlines were stressful.",
        "The examples used in class made learning easy.",
        "I felt the course could have been more interactive."
    ]
}
```

```python
df = pd.DataFrame(feedback_data)
df
```

```python
senti_p = pipeline("sentiment-analysis")
```

```python
summ_p = pipeline("summarization")
```

```python
all_feedbacks = df["feedback"].tolist()
all_feedbacks
```

```python
sentiments = senti_p(all_feedbacks)
sentiments
```

```python
all_labels = [s['label'] for s in sentiments]
all_labels
```

```python
all_scores = [s['score'] for s in sentiments]
all_scores
```

```python
df['Sentiment'] = all_labels
df['Score'] = all_scores

df
```

```python
combined_feedback = " ".join(df["feedback"].tolist())
combined_feedback
```

```python
summ_p(combined_feedback, max_length = 38, min_length = 30)
```

# Tokenizer

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
encoding = tokenizer("Huggingface is amazing!")
encoding
```

```python
tokenizer.convert_ids_to_tokens(encoding['input_ids'])
```

```python
encoding = tokenizer("Huggingface is amazing!", "I am enjoying all this")
encoding
```

```python
tokenizer(["Huggingface is amazing!", "I am enjoying all this", "Hi"], padding = "max_length",
          max_length = 6, truncation = True)
```

```python
tokenizer(["Huggingface is amazing!", "I am enjoying all this", "Hi"], padding = "max_length",
          max_length = 6, truncation = True, return_tensors = "pt")
```

```python
tokenizer(["Huggingface is amazing!", "I am enjoying all this", "Hi"], padding = "max_length",
          max_length = 6, truncation = True, return_tensors = "np")
```
