# Putting together

```python
from transformers import AutoTokenizer

model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)
```

```python
model.config
```

```python
sent = "I love Hugging face"
inputs = tokenizer(sent, return_tensors = "pt")
```

```python
inputs
```

```python
outputs = model(**inputs)
outputs
```

```python
d = {'a': 10, 'b': 20}

def fun(a, b):
  return a+b

print(fun(d['a'], d['b']))
print(fun(**d))
```

```python
outputs.logits[0]
```

```python
import torch
probs = torch.softmax(outputs.logits, dim = -1) # dim = 1
probs
```

```python
predicted_class = torch.argmax(probs, dim = -1)
predicted_class
```

```python
!pip install datasets

from datasets import load_dataset

fb_data = load_dataset("csv", data_files = "Student_feedback.csv")
fb_data
```

```python
# @title
import pandas as pd

feedback_data = {
    "feedback": [
        "The course content was very well structured and informative.",
        "The lectures were easy to understand.",
        "I really enjoyed the practical sessions and assignments.",
        "The instructor explained concepts clearly.",
        "The workload was heavy and deadlines were stressful.",
        "The examples used in class made learning easy.",
        "I felt the course could have been more interactive."
    ],
    # ADD THIS: Manual labels (1=positive, 0=negative)
    "labels": [1, 0, 1, 0, 0, 1, 0]
}

df = pd.DataFrame(feedback_data)

df.to_csv("Student_feedback.csv", index=False)
```

```python
fb_data = fb_data['train'].train_test_split(test_size = 0.25)
fb_data
```

```python
def tokenize_function(example):
  return tokenizer(
      example['feedback'],
      truncation = True,
      padding = "max_length"
  )
# tokenized_dataset = dataset.map(tokenize_function, batched = False)
fb_tokenized_dataset = fb_data.map(tokenize_function, batched = False)
fb_tokenized_dataset
```
